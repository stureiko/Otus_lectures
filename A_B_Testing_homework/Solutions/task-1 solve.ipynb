# Простой А/Б тест
## Условия
Маркетологи изменили рекламную компанию и наблюдаеют увеличение конверсии пользователей с 0.01% до 0.013%
Было собрано 100.000 наблюдений с аудитории до эксперимента и после эксперимента.
## Задание
1. Оценить есть ли статистически значимые различия в конверсии пользователей или мы можем считать это случайным эффектом?
2. Какой минимальный размер наблюдений необходимо набрать, чтобы сказать, что статистические различия есть?
3. Какой минимальный эффект на этих данных мы можем обнаружить?


```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest, proportion_effectsize
from statsmodels.stats.power import NormalIndPower
```

```python
# Заданные параметры
n_control = 100_000 
n_treatment = 100_000
p_control = 0.0010
p_treatment = 0.0013

# Считаем количество "успешных" событий (покупок)
success_control = int(n_control * p_control)
success_treatment = int(n_treatment * p_treatment)
```

```python
# Подставляем в z-тест
counts = [success_treatment, success_control]  # [эксперимент, контроль]
nobs = [n_treatment, n_control]

stat, pval = proportions_ztest(counts, nobs)

print(f"Control conversion: {success_control / n_control:.5f}")
print(f"Treatment conversion: {success_treatment / n_treatment:.5f}")
print(f"z-statistic: {stat:.3f}, p-value: {pval:.5f}")

# Интерпретация
alpha = 0.05
if pval < alpha:
    print("✅ Различие в конверсии статистически значимо.")
else:
    print("❌ Нет статистически значимого различия.")
```

```python
effect_size = proportion_effectsize(p_control, p_treatment)

# Параметры теста
alpha = 0.05       # Уровень значимости
power = 0.8        # Желаемая мощность теста

# Расчет необходимого размера выборки на группу
analysis = NormalIndPower()
sample_size = analysis.solve_power(effect_size=effect_size, power=power, alpha=alpha, ratio=1, alternative='two-sided')

print(f"🔢 Необходимое общее количество пользователей: {int(sample_size):,}")
```
