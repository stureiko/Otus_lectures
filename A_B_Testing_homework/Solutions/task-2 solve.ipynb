# Простой А/Б тест

## Условия

Маркетологи изменили рекламную компанию и наблюдаеют увеличение среднего чека с 2000 до 2010 рублей и дисперсией около 500 рублей.

Было собрано 100000 наблюдений с аудитории до эксперимента и после эксперимента.

## Задание

1. Оценить есть ли статистически значимые различия или мы можем считать это случайным эффектом?
2. Какой минимальный размер наблюдений необходимо набрать, чтобы сказать, что статистические различия есть?
3. Какой минимальный эффект на этих данных мы можем обнаружить?


```python
import numpy as np
import pandas as pd

from scipy.stats import ttest_ind
from statsmodels.stats.power import TTestIndPower
```

```python
df = pd.read_csv('ab_test_data.csv')
df.head()
```

```python
control = df[df['group'] == 'control']['check_amount']
test = df[df['group'] == 'test']['check_amount']

stat, pval = ttest_ind(test, control)
print(f"t-statistic: {stat:.3f}, p-value: {pval:.5f}")

# Интерпретация
alpha = 0.05
if pval < alpha:
    print("✅ Различие в конверсии статистически значимо.")
else:
    print("❌ Нет статистически значимого различия.")
```

```python
effect_size = (control.mean() - test.mean()) / control.std()

# Анализ мощности
analysis = TTestIndPower()
sample_size = analysis.solve_power(effect_size=effect_size,
                                   power=0.8,
                                   alpha=0.05,
                                   ratio=1.0,
                                   alternative='two-sided')

print(f"🔢 Необходимое количество наблюдений: {int(sample_size):,}")
```
