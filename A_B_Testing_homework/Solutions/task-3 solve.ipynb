## Условия
Маркетологи изменили рекламную компанию и наблюдаеют изменения в поведении пользователей

## Задание
1. Оценить есть ли статистически значимые различия или мы можем считать это случайным эффектом? (Ответ - изменения есть - покажите это)


```python
import numpy as np
import pandas as pd

from scipy.stats import ttest_ind
from statsmodels.stats.power import TTestIndPower
```

```python
df = pd.read_csv('ab_test_data.csv')
```

```python
control = df.loc[df['group'] == 'control', 'check_amount']
test = df.loc[df['group'] == 'test', 'check_amount']

stat, pval = ttest_ind(test, control)
print(f"t-statistic: {stat:.3f}, p-value: {pval:.5f}")

# Интерпретация
alpha = 0.05
if pval < alpha:
    print("✅ Различие в конверсии статистически значимо.")
else:
    print("❌ Нет статистически значимого различия.")
```

```python
control = df.loc[(df['group'] == 'control') & (df['age_group'] == "25-34"), 'check_amount']
test = df.loc[(df['group'] == 'test') & (df['age_group'] == "25-34"), 'check_amount']

stat, pval = ttest_ind(test, control)
print(f"t-statistic: {stat:.3f}, p-value: {pval:.5f}")

# Интерпретация
alpha = 0.05
if pval < alpha:
    print("✅ Различие в конверсии статистически значимо.")
else:
    print("❌ Нет статистически значимого различия.")
```
