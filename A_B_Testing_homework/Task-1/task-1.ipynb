# Простой А/Б тест

## Условия

Маркетологи изменили рекламную компанию и наблюдаеют увеличение конверсии пользователей с 0.01% до 0.013%
Было собрано 100000 наблюдений с аудитории до эксперимента и после эксперимента.

## Задание

1. Оценить есть ли статистически значимые различия в конверсии пользователей или мы можем считать это случайным эффектом?
2. Какой минимальный размер наблюдений необходимо набрать, чтобы сказать, что статистические различия есть?
3. Какой минимальный эффект на этих данных мы можем обнаружить?
