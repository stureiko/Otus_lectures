# Простой А/Б тест

## Условия

Маркетологи изменили рекламную компанию и наблюдаеют увеличение среднего чека с 2000 до 2010 рублей и дисперсией около 500 рублей.


## Задание

1. Оценить есть ли статистически значимые различия или мы можем считать это случайным эффектом?
2. Какой минимальный размер наблюдений необходимо набрать, чтобы сказать, что статистические различия есть?
3. Какой минимальный эффект на этих данных мы можем обнаружить?
