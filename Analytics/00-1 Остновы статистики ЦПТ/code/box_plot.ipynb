```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
m = np.random.uniform(0, 10, 1000)
```

```python
m[12] = 25
m[50] = 34
```

```python
plt.boxplot(m)
plt.show()
```

```python
# Вычисляем квартильные значения и IQR
q1 = np.percentile(m, 25)  # Первый квартиль
q3 = np.percentile(m, 75)  # Третий квартиль
iqr = q3 - q1                 # Межквартильный размах

# Определяем границы для выбросов
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

# Нахождение выбросов
outliers = [x for x in m if x < lower_bound or x > upper_bound]
print("Выбросы:", outliers)

# Нахождение индексов выбросов
outlier_indices = [index for index, value in enumerate(m) if value < lower_bound or value > upper_bound]
print("Индексы выбросов:", outlier_indices)
```
