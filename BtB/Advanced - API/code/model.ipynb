# Базовый ноутбук 
В этом ноутбуке МЛ инженер проводил эксперименты по формиррованию модели

код по формировани модели.

Для иллюстрации возможностей Pydantic модель принимает ряд параметров:
- float
- string произвольный
- string ограниченной длины
- string из фиксированного набора, как категория
- массив float открытой длины
- массив float фиксированной длины
