```python
from kafka import KafkaProducer
from json import dumps

# Конфигурация Kafka
topic = "input"
bootstrap_servers = "127.0.0.1:9092"

# Создание Kafka Producer
producer = KafkaProducer(
    bootstrap_servers=bootstrap_servers,
    value_serializer=lambda x: dumps(x).encode("utf-8"),
)
```

```python
# Данные для отправки
data = {
    "Pclass": 3,
    "Age": 22.0,
    "Fare": 7.2500
}

# Отправка данных в топик
producer.send(topic, value=data)
producer.flush()

print(f"Sent data to topic '{topic}': {data}")
```

```python
import random

def generate_random_passenger():
    # Генерация Pclass (1, 2 или 3)
    pclass = random.choice([1, 2, 3])
    
    # Генерация Age (от 0 до 80)
    age = round(random.uniform(0, 80))
    
    # Генерация Fare в зависимости от Pclass
    if pclass == 1:
        fare = round(random.uniform(50, 500), 2)  # Высокие цены для первого класса
    elif pclass == 2:
        fare = round(random.uniform(10, 50), 2)   # Средние цены для второго класса
    else:  # Pclass == 3
        fare = round(random.uniform(0, 20), 2)    # Низкие цены для третьего класса
    
    # Возвращаем сгенерированные данные
    return {
        "Pclass": pclass,
        "Age": age,
        "Fare": fare
    }

# Пример использования
random_passenger = generate_random_passenger()
print("Generated Passenger Data:", random_passenger)
```

```python
# Пример использования
random_passenger = generate_random_passenger()
# Отправка данных в топик
producer.send(topic, value=random_passenger)
producer.flush()

print(f"Sent data to topic '{topic}': {random_passenger}")
```
