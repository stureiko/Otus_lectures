```python
import requests

import apimoex
import pandas as pd
```

```python
# Создаём сессию
with requests.Session() as session:
    # Получаем исторические свечи (OHLC) для индекса IMOEX
    data = apimoex.get_board_candles(
        session,
        security='IMOEX',         # тикер индекса
        market='index',           # указываем, что это индекс
        interval=24,              # 24 = дневные свечи
        start='2024-01-01',       # начальная дата
        end='2024-12-31'          # конечная дата
    )

# Преобразуем в DataFrame
df = pd.DataFrame(data)
```
