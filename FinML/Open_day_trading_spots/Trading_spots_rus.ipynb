Working with official Moscow Exchange (MOEX) API

### Основные эндпоинты:
База: https://iss.moex.com/iss

Текущие данные: https://iss.moex.com/iss/engines/stock/markets/shares/boards/TQBR/securities/{TICKER}.json

История: https://iss.moex.com/iss/history/engines/stock/markets/shares/boards/TQBR/securities/{TICKER}.json

Дивиденды: https://iss.moex.com/iss/securities/{TICKER}/dividends.json

Информация: https://iss.moex.com/iss/securities/{TICKER}.json

Индексы: https://iss.moex.com/iss/engines/stock/markets/index/securities.json

```python
import requests
import pandas as pd
from datetime import datetime, timedelta
import json
import time
```

```python
# Базовый URL API MOEX
MOEX_API_BASE = "https://iss.moex.com/iss"

# Популярные акции MOEX
POPULAR_STOCKS = {
    'SBER': 'Сбербанк',
    'GAZP': 'Газпром',
    'LKOH': 'Лукойл',
    'GMKN': 'Норникель',
    'YNDX': 'Яндекс',
    'NVTK': 'Новатэк',
    'ROSN': 'Роснефть',
    'TATN': 'Татнефть',
    'MTSS': 'МТС',
    'VTBR': 'ВТБ'
}
```

```python
def get_current_trading_data(ticker, verbose=False):
    """
    Получить текущие торговые данные по акции
    Get current trading data for a stock
    
    Возвращает: цену, объем, изменение и т.д.
    """
    if verbose:
        print(f"\n{'='*80}")
        print(f"ТЕКУЩИЕ ТОРГОВЫЕ ДАННЫЕ: {ticker}")
        print(f"CURRENT TRADING DATA: {ticker}")
        print(f"{'='*80}")
    
    # URL для получения данных по акции на основном рынке (TQBR - Т+2)
    url = f"{MOEX_API_BASE}/engines/stock/markets/shares/boards/TQBR/securities/{ticker}.json"
    
    try:
        # Делаем запрос
        response = requests.get(url)
        response.raise_for_status()  # Проверка на ошибки HTTP
        
        data = response.json()
        
        # Извлекаем данные из ответа
        # MOEX API возвращает данные в формате с метаданными и данными
        securities = data.get('securities', {})
        marketdata = data.get('marketdata', {})
        
        if not securities.get('data') or not marketdata.get('data'):
            print(f" Данные для {ticker} не найдены")
            return None
        
        # Преобразуем в DataFrame для удобства
        securities_df = pd.DataFrame(
            securities['data'],
            columns=securities['columns']
        )
        
        marketdata_df = pd.DataFrame(
            marketdata['data'],
            columns=marketdata['columns']
        )
        
        # Объединяем данные
        if len(securities_df) > 0 and len(marketdata_df) > 0:
            sec_data = securities_df.iloc[0]
            market_data = marketdata_df.iloc[0]
            
            result = {
                'ticker': ticker,
                'name': sec_data.get('SHORTNAME', 'N/A'),
                'last_price': market_data.get('LAST', 0),
                'prev_price': market_data.get('PREVPRICE', 0),
                'change': market_data.get('CHANGE', 0),
                'change_percent': market_data.get('CHANGEPERCENT', 0),
                'open': market_data.get('OPEN', 0),
                'high': market_data.get('HIGH', 0),
                'low': market_data.get('LOW', 0),
                'volume': market_data.get('VOLUME', 0),
                'value': market_data.get('VALUE', 0),  # Объем в рублях
                'update_time': market_data.get('UPDATETIME', 'N/A'),
                'lot_size': sec_data.get('LOTSIZE', 1)
            }
            
            # Вывод результатов
            if verbose:
                print(f"\nКомпания: {result['name']}")
                print(f"\nЦена / Price:")
                print(f"  Текущая / Current: {result['last_price']:.2f} ₽")
                print(f"  Открытие / Open: {result['open']:.2f} ₽")
                print(f"  Максимум / High: {result['high']:.2f} ₽")
                print(f"  Минимум / Low: {result['low']:.2f} ₽")
                
                print(f"\nИзменение / Change:")
                print(f"  Абсолютное / Absolute: {result['change']:+.2f} ₽")
                print(f"  Процентное / Percent: {result['change_percent']:+.2f}%")
                
                print(f"\nОбъем / Volume:")
                print(f"  Количество акций / Shares: {result['volume']:,.0f}")
                print(f"  Оборот / Turnover: {result['value']:,.0f} ₽")
                
                print(f"\nОбновлено / Updated: {result['update_time']}")
                print(f"Размер лота / Lot size: {result['lot_size']}")
            
            return result
        else:
            print(f"Нет данных для {ticker}")
            return None
            
    except requests.exceptions.RequestException as e:
        print(f"Ошибка запроса / Request error: {e}")
        return None
    except Exception as e:
        print(f"Ошибка обработки данных / Data processing error: {e}")
        return None
```

```python
def get_market_cap(ticker, verbose=False):
    """
    Получить рыночную капитализацию акции
    Get market capitalization of a stock
    """
    if verbose:
        print(f"\n{'='*80}")
        print(f"КАПИТАЛИЗАЦИЯ: {ticker}")
        print(f"{'='*80}")
    
    # URL для получения информации о ценной бумаге
    url = f"{MOEX_API_BASE}/securities/{ticker}.json"
    
    try:
        response = requests.get(url)
        response.raise_for_status()
        
        data = response.json()
        
        # Ищем капитализацию в различных разделах
        description = data.get('description', {})
        
        if description.get('data'):
            desc_df = pd.DataFrame(
                description['data'],
                columns=description['columns']
            )
            
            # Найти капитализацию
            market_cap_row = desc_df[desc_df['name'] == 'MARKETPRICE']
            issue_cap_row = desc_df[desc_df['name'] == 'ISSUECAPITALIZATION']
            
            market_cap = None
            if not market_cap_row.empty:
                market_cap = market_cap_row.iloc[0]['value']
            elif not issue_cap_row.empty:
                market_cap = issue_cap_row.iloc[0]['value']
            
            # Получить количество акций
            issued_shares_row = desc_df[desc_df['name'] == 'ISSUESIZE']
            issued_shares = None
            if not issued_shares_row.empty:
                issued_shares = issued_shares_row.iloc[0]['value']
            
            # Получить текущую цену для расчета
            trading_data = get_current_trading_data(ticker)
            
            result = {
                'ticker': ticker,
                'market_cap': market_cap,
                'issued_shares': issued_shares,
                'current_price': trading_data['last_price'] if trading_data else None
            }
            
            # Рассчитать капитализацию если есть данные
            if issued_shares and result['current_price']:
                try:
                    shares = float(issued_shares)
                    calculated_cap = shares * result['current_price']
                    result['calculated_cap'] = calculated_cap
                except:
                    pass
            
            # Вывод результатов
            if verbose:
                print(f"\nКапитализация / Market Cap:")
                
                if market_cap:
                    try:
                        cap_value = float(market_cap)
                        print(f"  Официальная / Official: {cap_value:,.0f} ₽")
                        print(f"  = {cap_value/1e9:.2f} млрд ₽ / billion RUB")
                    except:
                        print(f"  Официальная / Official: {market_cap}")
                
                if issued_shares:
                    print(f"\nАкций в обращении / Shares outstanding:")
                    try:
                        shares_value = float(issued_shares)
                        print(f"  {shares_value:,.0f} шт")
                    except:
                        print(f"  {issued_shares}")
                
                if result.get('calculated_cap'):
                    print(f"\nРасчетная капитализация / Calculated cap:")
                    print(f"  (Акции × Цена): {result['calculated_cap']:,.0f} ₽")
                    print(f"  = {result['calculated_cap']/1e9:.2f} млрд ₽ / billion RUB")
            
            return result
        else:
            print(f"Данные о капитализации не найдены")
            return None
            
    except requests.exceptions.RequestException as e:
        print(f"Ошибка запроса / Request error: {e}")
        return None
    except Exception as e:
        print(f"Ошибка обработки данных / Data processing error: {e}")
        return None
```

```python
def get_dividends(ticker, verbose=False):
    """
    Получить информацию о дивидендах
    Get dividend information
    """
    if verbose:
        print(f"\n{'='*80}")
        print(f"ДИВИДЕНДЫ: {ticker}")
        print(f"{'='*80}")
    
    # URL для получения дивидендов
    url = f"{MOEX_API_BASE}/securities/{ticker}/dividends.json"
    
    try:
        response = requests.get(url)
        response.raise_for_status()
        
        data = response.json()
        
        dividends = data.get('dividends', {})
        
        if not dividends.get('data'):
            print(f"Информация о дивидендах не найдена")
            return None
        
        # Преобразуем в DataFrame
        div_df = pd.DataFrame(
            dividends['data'],
            columns=dividends['columns']
        )
        
        if len(div_df) == 0:
            print(f"Дивиденды не найдены")
            return None
        
        # Сортируем по дате (последние сверху)
        div_df = div_df.sort_values('registryclosedate', ascending=False)
        
        if verbose:
            print(f"\nНайдено {len(div_df)} выплат дивидендов")
            
            print(f"\nПоследние 10 выплат / Last 10 payments:")
            print("-" * 80)
            
            for idx, row in div_df.head(10).iterrows():
                record_date = row.get('registryclosedate', 'N/A')
                value = row.get('value', 0)
                currency = row.get('currencyid', 'RUB')
                
                print(f"Дата закрытия реестра / Record date: {record_date}")
                print(f"  Сумма / Amount: {value} {currency}")
                print("-" * 80)
            
            # Статистика
            total_div = div_df['value'].sum()
            avg_div = div_df['value'].mean()
            last_div = div_df.iloc[0]['value'] if len(div_df) > 0 else 0
            
            print(f"\nСтатистика / Statistics:")
            print(f"  Последняя выплата / Last payment: {last_div:.2f} ₽")
            print(f"  Средняя выплата / Average payment: {avg_div:.2f} ₽")
            print(f"  Всего выплачено / Total paid: {total_div:.2f} ₽")
            
            # Рассчитать дивидендную доходность
            trading_data = get_current_trading_data(ticker)
            if trading_data and trading_data['last_price'] > 0:
                # Берем сумму дивидендов за последний год
                one_year_ago = datetime.now() - timedelta(days=365)
                div_df['date'] = pd.to_datetime(div_df['registryclosedate'])
                recent_divs = div_df[div_df['date'] >= one_year_ago]
                annual_div = recent_divs['value'].sum()
                
                if annual_div > 0:
                    div_yield = (annual_div / trading_data['last_price']) * 100
                    print(f"\n💹 Дивидендная доходность (год) / Dividend Yield (annual):")
                    print(f"  {div_yield:.2f}%")
                    print(f"  (Дивиденды за год / Текущая цена)")
        
        return div_df
        
    except requests.exceptions.RequestException as e:
        print(f"Ошибка запроса / Request error: {e}")
        return None
    except Exception as e:
        print(f"Ошибка обработки данных / Data processing error: {e}")
        return None
```

```python
def get_historical_data(ticker, days=30):
    """
    Получить исторические данные для расчета доходности
    Get historical data to calculate returns
    """
    # Рассчитываем даты
    end_date = datetime.now()
    start_date = end_date - timedelta(days=days+10)  # +10 для запаса
    
    # Форматируем даты для API
    from_date = start_date.strftime('%Y-%m-%d')
    till_date = end_date.strftime('%Y-%m-%d')
    
    # URL для исторических данных
    url = f"{MOEX_API_BASE}/history/engines/stock/markets/shares/boards/TQBR/securities/{ticker}.json"
    params = {
        'from': from_date,
        'till': till_date
    }
    
    try:
        response = requests.get(url, params=params)
        response.raise_for_status()
        
        data = response.json()
        
        history = data.get('history', {})
        
        if not history.get('data'):
            return None
        
        # Преобразуем в DataFrame
        hist_df = pd.DataFrame(
            history['data'],
            columns=history['columns']
        )
        
        # Фильтруем только нужные колонки
        hist_df = hist_df[['TRADEDATE', 'CLOSE', 'VOLUME', 'VALUE']].copy()
        hist_df['TRADEDATE'] = pd.to_datetime(hist_df['TRADEDATE'])
        
        # Сортируем по дате
        hist_df = hist_df.sort_values('TRADEDATE')
        
        # Убираем дни без торгов
        hist_df = hist_df[hist_df['CLOSE'].notna()]
        hist_df = hist_df[hist_df['CLOSE'] > 0]
        
        return hist_df
        
    except Exception as e:
        print(f"Ошибка получения истории для {ticker}: {e}")
        return None
```

```python
def compare_stocks_returns(tickers, period_days=30, verbose=False):
    """
    Сравнить несколько акций по доходности за период
    Compare multiple stocks by returns over a period
    """
    if verbose:
        print(f"\n{'='*80}")
        print(f"СРАВНЕНИЕ АКЦИЙ ПО ДОХОДНОСТИ ЗА {period_days} ДНЕЙ")
        print(f"{'='*80}")
    
    results = []
    
    for ticker in tickers:
        print(f"\nОбработка {ticker}...")
        
        try:
            # Получить исторические данные
            hist_df = get_historical_data(ticker, period_days)
            
            if hist_df is None or len(hist_df) < 2:
                print(f"Недостаточно данных для {ticker}")
                continue
            
            # Получить цены
            start_price = hist_df.iloc[0]['CLOSE']
            end_price = hist_df.iloc[-1]['CLOSE']
            
            # Рассчитать доходность
            returns = ((end_price - start_price) / start_price) * 100
            
            # Рассчитать волатильность
            hist_df['returns'] = hist_df['CLOSE'].pct_change()
            volatility = hist_df['returns'].std() * (252 ** 0.5) * 100
            
            # Средний дневной объем
            avg_volume = hist_df['VOLUME'].mean()
            avg_value = hist_df['VALUE'].mean()
            
            results.append({
                'Тикер / Ticker': ticker,
                'Компания / Company': POPULAR_STOCKS.get(ticker, ticker),
                'Нач. цена / Start': f"{start_price:.2f} ₽",
                'Кон. цена / End': f"{end_price:.2f} ₽",
                'Доходность % / Returns %': f"{returns:+.2f}",
                'Волатильность % / Volatility %': f"{volatility:.1f}",
                'Ср. объем / Avg Volume': f"{avg_volume:,.0f}",
                'Дней / Days': len(hist_df)
            })
            
            print(f"{ticker}: {returns:+.2f}%")
            
        except Exception as e:
            print(f"Ошибка для {ticker}: {e}")
    
    if not results:
        print("\nНе удалось получить данные ни по одной акции")
        return None
    
    # Создаем DataFrame для красивого вывода
    df = pd.DataFrame(results)
    
    # Сортируем по доходности (от большей к меньшей)
    df['_returns_num'] = df['Доходность % / Returns %'].str.replace('%', '').astype(float)
    df = df.sort_values('_returns_num', ascending=False)
    df = df.drop('_returns_num', axis=1)
    
    if verbose:
        print(f"\n{'='*80}")
        print("ИТОГОВАЯ ТАБЛИЦА / SUMMARY TABLE")
        print(f"{'='*80}")
        print(df.to_string(index=False))
        print(f"{'='*80}")
    
    # Определить лидеров и аутсайдеров
    best = results[0] if results else None
    worst = results[-1] if len(results) > 1 else None
    
    if verbose:
        if best:
            print(f"\nЛидер роста / Top Performer:")
            print(f"    {best['Тикер / Ticker']}: {best['Доходность % / Returns %']}%")
        
        if worst and len(results) > 1:
            print(f"\nМаксимальное падение / Worst Performer:")
            print(f"   {worst['Тикер / Ticker']}: {worst['Доходность % / Returns %']}%")
    
    return df
```

```python
def get_current_price(symbol, category='spot', verbose=False):
    """
    Получить текущую цену криптовалюты
    Get current cryptocurrency price
    
    Args:
        symbol: Торговая пара (например, 'BTCUSDT')
        category: Тип рынка ('spot', 'linear', 'inverse')
    
    Returns:
        dict с текущими данными
    """
    if verbose:
        print(f"\n{'='*80}")
        print(f"💰 ТЕКУЩАЯ ЦЕНА: {symbol}")
        print(f"{'='*80}")
    
    # URL для получения тикера
    url = f"{BYBIT_API_BASE}/v5/market/tickers"
    params = {
        'category': category,
        'symbol': symbol
    }
    
    try:
        response = requests.get(url, params=params)
        response.raise_for_status()
        
        data = response.json()
        
        if data['retCode'] != 0:
            print(f"Ошибка API: {data['retMsg']}")
            return None
        
        # Извлекаем данные
        result = data['result']['list'][0] if data['result']['list'] else None
        
        if not result:
            print(f"Данные для {symbol} не найдены")
            return None
        
        # Форматируем результат
        ticker_data = {
            'symbol': result['symbol'],
            'last_price': float(result['lastPrice']),
            'bid_price': float(result['bid1Price']) if result.get('bid1Price') else None,
            'ask_price': float(result['ask1Price']) if result.get('ask1Price') else None,
            'high_24h': float(result['highPrice24h']),
            'low_24h': float(result['lowPrice24h']),
            'volume_24h': float(result['volume24h']),
            'turnover_24h': float(result['turnover24h']),
            'price_change_24h': float(result['price24hPcnt']) * 100,  # в процентах
            'prev_price_24h': float(result['prevPrice24h'])
        }
        
        # Вывод результатов
        if verbose:
            print(f"\nТорговая пара: {ticker_data['symbol']}")
            
            print(f"\nЦена / Price:")
            print(f"  Последняя / Last: ${ticker_data['last_price']:,.2f}")
            print(f"  Bid: ${ticker_data['bid_price']:,.2f}" if ticker_data['bid_price'] else "  Bid: N/A")
            print(f"  Ask: ${ticker_data['ask_price']:,.2f}" if ticker_data['ask_price'] else "  Ask: N/A")
            
            print(f"\n24h Статистика / 24h Stats:")
            print(f"  Максимум / High: ${ticker_data['high_24h']:,.2f}")
            print(f"  Минимум / Low: ${ticker_data['low_24h']:,.2f}")
            print(f"  Изменение / Change: {ticker_data['price_change_24h']:+.2f}%")
            
            print(f"\nОбъем торгов / Trading Volume:")
            print(f"  Объем (24h): {ticker_data['volume_24h']:,.2f} {symbol.replace('USDT', '')}")
            print(f"  Оборот (24h): ${ticker_data['turnover_24h']:,.2f}")
        
        return ticker_data
        
    except requests.exceptions.RequestException as e:
        print(f"Ошибка запроса / Request error: {e}")
        return None
    except Exception as e:
        print(f"Ошибка обработки данных / Data processing error: {e}")
        return None
```

```python
def get_klines(symbol, interval='D', limit=200, category='spot', verbose=False):
    """
    Получить исторические данные (свечи)
    Get historical data (candlesticks/klines)
    
    Args:
        symbol: Торговая пара (например, 'BTCUSDT')
        interval: Интервал ('1', '3', '5', '15', '30', '60', '120', '240', 'D', 'W', 'M')
                  1 = 1 минута, D = 1 день, W = 1 неделя, M = 1 месяц
        limit: Количество свечей (макс. 1000)
        category: Тип рынка ('spot', 'linear', 'inverse')
    
    Returns:
        DataFrame с историческими данными
    """
    if verbose:
        print(f"\n{'='*80}")
        print(f"ИСТОРИЧЕСКИЕ ДАННЫЕ: {symbol}")
        print(f"{'='*80}")
    
    # URL для получения свечей
    url = f"{BYBIT_API_BASE}/v5/market/kline"
    
    # Рассчитать время начала (ByBit требует timestamp в миллисекундах)
    end_time = int(time.time() * 1000)
    
    params = {
        'category': category,
        'symbol': symbol,
        'interval': interval,
        'limit': limit,
        'end': end_time
    }
    
    try:
        response = requests.get(url, params=params)
        response.raise_for_status()
        
        data = response.json()
        
        if data['retCode'] != 0:
            print(f"Ошибка API: {data['retMsg']}")
            return None
        
        # Извлекаем свечи
        klines = data['result']['list']
        
        if not klines:
            print(f"Исторические данные не найдены")
            return None
        
        # Создаем DataFrame
        df = pd.DataFrame(klines, columns=[
            'timestamp', 'open', 'high', 'low', 'close', 'volume', 'turnover'
        ])
        
        # Конвертируем типы данных
        df['timestamp'] = pd.to_datetime(df['timestamp'].astype(float), unit='ms')
        df['open'] = df['open'].astype(float)
        df['high'] = df['high'].astype(float)
        df['low'] = df['low'].astype(float)
        df['close'] = df['close'].astype(float)
        df['volume'] = df['volume'].astype(float)
        df['turnover'] = df['turnover'].astype(float)
        
        # Сортируем по времени (от старых к новым)
        df = df.sort_values('timestamp').reset_index(drop=True)
        
        # Устанавливаем timestamp как индекс
        df.set_index('timestamp', inplace=True)
        
        if verbose:
            print(f"\nЗагружено {len(df)} свечей")
            print(f"Период: {df.index[0].strftime('%Y-%m-%d %H:%M')} - {df.index[-1].strftime('%Y-%m-%d %H:%M')}")
            print(f"Интервал: {interval}")
            
            print(f"\nПоследние 5 свечей / Last 5 candles:")
            print(df[['open', 'high', 'low', 'close', 'volume']].tail().to_string())
            
        return df
        
    except requests.exceptions.RequestException as e:
        print(f"Ошибка запроса / Request error: {e}")
        return None
    except Exception as e:
        print(f"Ошибка обработки данных / Data processing error: {e}")
        return None
```

```python
def calculate_returns(symbol, days=30, interval='D', category='spot', verbose=False):
    """
    Рассчитать доходность за период
    Calculate returns over a period
    
    Args:
        symbol: Торговая пара
        days: Количество дней
        interval: Интервал свечей
        category: Тип рынка
    
    Returns:
        dict с результатами
    """
    if verbose:
        print(f"\n{'='*80}")
        print(f"ДОХОДНОСТЬ ЗА {days} ДНЕЙ: {symbol}")
        print(f"{'='*80}")
        
    # Получить исторические данные
    df = get_klines(symbol, interval=interval, limit=days+10, category=category)
    
    if df is None or len(df) < 2:
        print(f"Недостаточно данных для расчета")
        return None
    
    # Взять нужное количество дней
    if len(df) > days:
        df = df.tail(days)
    
    # Рассчитать доходность
    start_price = df['close'].iloc[0]
    end_price = df['close'].iloc[-1]
    returns = ((end_price - start_price) / start_price) * 100
    
    # Рассчитать волатильность
    df['returns'] = df['close'].pct_change()
    volatility = df['returns'].std()
    
    # Рассчитать максимальную просадку
    df['cumulative'] = (1 + df['returns']).cumprod()
    df['cummax'] = df['cumulative'].cummax()
    df['drawdown'] = (df['cumulative'] - df['cummax']) / df['cummax']
    max_drawdown = df['drawdown'].min() * 100
    
    # Средний объем
    avg_volume = df['volume'].mean()
    avg_turnover = df['turnover'].mean()
    
    result = {
        'symbol': symbol,
        'days': len(df),
        'start_price': start_price,
        'end_price': end_price,
        'returns': returns,
        'volatility': volatility * 100,
        'max_drawdown': max_drawdown,
        'avg_volume': avg_volume,
        'avg_turnover': avg_turnover
    }
    
    # Вывод результатов
    if verbose:
        print(f"\nАнализ за {len(df)} дней:")
        print(f"  Начальная цена / Start: ${start_price:,.2f}")
        print(f"  Конечная цена / End: ${end_price:,.2f}")
        print(f"  Доходность / Returns: {returns:+.2f}%")
        
        print(f"\nРиск-метрики / Risk Metrics:")
        print(f"  Волатильность / Volatility: {volatility*100:.2f}%")
        print(f"  Макс. просадка / Max Drawdown: {max_drawdown:.2f}%")
        
        print(f"\nСредние показатели / Average Metrics:")
        print(f"  Объем / Volume: {avg_volume:,.2f}")
        print(f"  Оборот / Turnover: ${avg_turnover:,.2f}")
    
    return result
```

```python
def compare_cryptos(symbols, days=30, category='spot', verbose=False):
    """
    Сравнить несколько криптовалют по доходности
    Compare multiple cryptocurrencies by returns
    
    Args:
        symbols: Список торговых пар
        days: Количество дней для анализа
        category: Тип рынка
    
    Returns:
        DataFrame со сравнением
    """
    if verbose:
        print(f"\n{'='*80}")
        print(f"СРАВНЕНИЕ КРИПТОВАЛЮТ ЗА {days} ДНЕЙ")
        print(f"{'='*80}")
    
    results = []
    
    for symbol in symbols:
        print(f"\nОбработка {symbol}...")
        
        try:
            # Получить данные
            df = get_klines(symbol, interval='D', limit=days+10, category=category)
            
            if df is None or len(df) < 2:
                print(f"  Недостаточно данных для {symbol}")
                continue
            
            # Взять нужное количество дней
            if len(df) > days:
                df = df.tail(days)
            
            # Рассчитать метрики
            start_price = df['close'].iloc[0]
            end_price = df['close'].iloc[-1]
            returns = ((end_price - start_price) / start_price) * 100
            
            df['returns'] = df['close'].pct_change()
            volatility = df['returns'].std() * 100
            
            avg_volume = df['volume'].mean()
            
            # Получить текущую цену
            current = get_current_price(symbol, category)
            current_price = current['last_price'] if current else end_price
            change_24h = current['price_change_24h'] if current else 0
            
            results.append({
                'Символ / Symbol': symbol,
                'Название / Name': POPULAR_PAIRS.get(symbol, symbol)[:20],
                'Цена / Price': f"${current_price:,.2f}",
                'Изм 24h % / 24h Chg': f"{change_24h:+.2f}",
                f'Доходность {days}д % / {days}d Return': f"{returns:+.2f}",
                'Волатильность % / Volatility': f"{volatility:.1f}",
                'Ср. объем / Avg Vol': f"{avg_volume:,.0f}",
                'Дней / Days': len(df)
            })
            
            print(f"  {symbol}: {returns:+.2f}%")
            
        except Exception as e:
            print(f"Ошибка для {symbol}: {e}")
    
    if not results:
        print("\nНе удалось получить данные ни по одной паре")
        return None
    
    # Создаем DataFrame
    df = pd.DataFrame(results)
    
    # Сортируем по доходности
    df['_returns_num'] = df[f'Доходность {days}д % / {days}d Return'].str.replace('%', '').astype(float)
    df = df.sort_values('_returns_num', ascending=False)
    df = df.drop('_returns_num', axis=1)
    
    if verbose:
        print(f"\n{'='*80}")
        print("ИТОГОВАЯ ТАБЛИЦА / SUMMARY TABLE")
        print(f"{'='*80}")
        print(df.to_string(index=False))
        print(f"{'='*80}")
        
        # Определить лидеров
        if len(results) > 0:
            best = results[0]
            print(f"\nЛидер роста / Top Performer:")
            print(f"   {best['Символ / Symbol']}: {best[f'Доходность {days}д % / {days}d Return']}%")
        
        if len(results) > 1:
            worst = results[-1]
            print(f"\nХудший результат / Worst Performer:")
            print(f"   {worst['Символ / Symbol']}: {worst[f'Доходность {days}д % / {days}d Return']}%")
    
    return df
```
