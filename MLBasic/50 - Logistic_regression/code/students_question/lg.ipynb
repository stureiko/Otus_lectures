```python
from feature_engine.selection import DropCorrelatedFeatures
from sklearn import datasets
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
def train_and_score(data, results):
    estimator = LogisticRegression()
    scaler = StandardScaler()
    train, test, train_result, test_result = train_test_split(data, results, train_size=0.7, stratify=results, random_state=42)
    estimator.fit(scaler.fit_transform(train), train_result)
    return f1_score(estimator.predict(scaler.transform(test)), test_result)
```

```python

(all_ft, categories) = datasets.load_breast_cancer(return_X_y=True, as_frame=True)
reduced_ft = DropCorrelatedFeatures(threshold=0.85).fit_transform(all_ft)
```

```python

print(f'''All features F1: {all_ft.shape}: {train_and_score(all_ft, categories)}''')
print(f'''Reduced features F1: {reduced_ft.shape}: {train_and_score(reduced_ft, categories)}''')
```
