## T-test (для двух выборок).
Самый важный!

<div class="alert alert-info">
<b>Гипотеза: </b>

Правда ли, что инженеры (Engineer) зарабатывают в среднем больше, чем люди из сферы развлечений (Entertainment)?

```python
from scipy import stats  # Импортируем библиотеку


Engineer_salary = data[data['profession']=='Engineer']['salary']
Entertainment_salary = data[data['profession']=='Entertainment']['salary']

```

```python
Engineer_salary.shape[0], Entertainment_salary.shape[0]
```

<div class="alert alert-info">
<b>Описание: </b>
Такой тест показывает, имеют ли две выборки данных разные средние значения (Теперь у нас две разные выборки, а не одна!). Здесь мы принимаем нулевую гипотезу о том, что обе группы имеют равные средние. Для этого нам не нужен известный параметр численности.
</div>

```python
plt.figure(figsize=(10, 6))
plt.hist(Engineer_salary, bins=30, density=True, alpha=0.6, color='skyblue', label='Engineer_salary')
plt.hist(Entertainment_salary, bins=30, density=True, alpha=0.6, color='darkblue', label='Entertainment_salary')

# Построение кривой нормального распределения
x = np.linspace(Engineer_salary.min(), Engineer_salary.max(), 1000)
pdf = stats.norm.pdf(x, Engineer_salary.mean(), Engineer_salary.std())
plt.plot(x, pdf, 'r-', label='Нормальное распределение Engineer_salary')

x = np.linspace(Entertainment_salary.min(), Entertainment_salary.max(), 1000)
pdf = stats.norm.pdf(x, Entertainment_salary.mean(), Entertainment_salary.std())
plt.plot(x, pdf, 'r-', label='Нормальное распределение Entertainment_salary')

plt.xlabel('Зарплата')
plt.ylabel('Плотность')
plt.legend()
plt.grid()
plt.show()
```

## Проверим наши выборки на нормальность распределения:
1. Тест Шапиро-Уилка (Shapiro-Wilk test)
- Один из самых популярных тестов.
- Подходит для малых и средних выборок (до 5000 наблюдений).

2. Тест Колмогорова-Смирнова (K-S test)
- Подходит для средних и больших выборок.
- Сравнивает эмпирическую функцию распределения с нормальным распределением.

```python
stat, p_value = stats.shapiro(Engineer_salary)
print('Тест Шапиро-Уилка')
print(f'p-value = {round(p_value, 8)}')
# Интерпретация результата
alpha = 0.05
if p_value < alpha:
    print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: данные отклоняются от нормального распределения.{bcolors.ENDC}")
else:
    print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: данные распределены нормально.{bcolors.ENDC}")
```

```python
stat, p_value = stats.kstest(
    Engineer_salary,
    'norm'
)

print('Тест Колмогорова-Смирнова')
print(f'p-value = {round(p_value, 8)}')

# Интерпретация результата
alpha = 0.05
if p_value < alpha:
    print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: данные отклоняются от нормального распределения.{bcolors.ENDC}")
else:
    print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: данные распределены нормально.{bcolors.ENDC}")
```

<div class="alert alert-info">
<b>Замечание: </b>
Для того, чтобы воспользоваться двухвыборочным t-тестом для сравнения средних нам необходимо убедиться, что дисперсии наших выборок равны.

Для этого можно использовать следующие методы:
- Бартлетта. Он чувствителен к нормальности распределения;
- Левена (Ливиня). Менее чувствителен к нормальности, самый популярный тест;
- Флигнера-Килина. Не чувствителен к отклонениям от нормальности и выбросам, подходит для ненормального распределения;

Будем использовать тест Левена
</div>

```python
# Применение теста Левена
stat, p_value = stats.levene(Engineer_salary, Entertainment_salary)

# Вывод результатов
print(f"Статистика Левена: {np.round(stat, 4)}")
print(f"p-значение: {np.round(p_value, 4)}")

# Интерпретация результата
alpha = 0.05
if p_value < alpha:
    print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: дисперсии не равны.{bcolors.ENDC}")
else:
    print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: дисперсии равны.{bcolors.ENDC}")
```

## Приведение к нормальному виду

### Преобразование Бокса-Кокса

```python
Engineer_transformed, Engineer_fitted_lambda = stats.boxcox(Engineer_salary)
Entertainment_transformed, Entertainment_fitted_lambda = stats.boxcox(Entertainment_salary)
```

```python
stat, p_value = stats.shapiro(Engineer_transformed)
print('Тест Шапиро-Уилка')
print(f'p-value = {round(p_value, 8)}')
# Интерпретация результата
alpha = 0.05
if p_value < alpha:
    print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: данные отклоняются от нормального распределения.{bcolors.ENDC}")
else:
    print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: данные распределены нормально.{bcolors.ENDC}")
```

## Критерий Стьюдента

```python
# Так как распределение похоже на нормальное, то берем Критерий Стьюдента

t, p_value  = stats.ttest_ind(Engineer_transformed, Entertainment_transformed) # Критерий Стьюдента (t-test)

print('Критерий Стьюдента')
print(f'p-value = {p_value}')
# Интерпретация результата
alpha = 0.05
if p_value < alpha:
    print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: средние в выборках различаются.{bcolors.ENDC}")
else:
    print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: средние равны.{bcolors.ENDC}")
```

<div class="alert alert-success">


Мы получаем хорошее значение p, равное 0.0, и, таким образом, мы отклоняем нулевую гипотезу, то есть - средние значения двух распределений различны и это различие статистически значимо

## Дисперсионный анализ

Функция scipy.stats.f_oneway() выполняет однофакторный дисперсионный анализ (ANOVA, Analysis of Variance).

Однофакторный ANOVA используется для проверки, есть ли статистически значимые различия между средними значениями более чем двух независимых групп.

### Формулировка гипотез:
- H0 (нулевая гипотеза): Средние значения всех групп равны.
- H1 (альтернативная гипотеза): Хотя бы одна группа имеет среднее, отличающееся от остальных.

```python
from scipy.stats import f_oneway # Импортирем функцию для дисперсионного анализа из библиотеки scipy пакета stats


F, p_value = f_oneway(data[data["profession"] == 'Engineer']["salary"],
                data[data["profession"] == 'Doctor']["salary"],
                data[data["profession"] == 'Marketing']["salary"],
                data[data["profession"] == 'Lawyer']["salary"],
                data[data["profession"] == 'Artist']["salary"],
                data[data["profession"] == 'Homemaker']["salary"]) 
# Дисперсионный анализ (ДА) в python

alpha = 0.05  # Уровень значимости
print(F, p_value)   # Смотрим на результаты ДА

if p_value < alpha:
    print('Одинаковое распределение (не отвергаем H0)')
else:
    print('Разное распределение (отклоняем H0)')
```

## Критерий Манна-Уитни

непараметрический тест, который применяется для проверки гипотезы о том, что два независимых выборки происходят из одного и того же распределения.

### Когда используется:
1.	Сравнение двух независимых выборок.
2.	Когда данные не соответствуют нормальному распределению.
3.	Когда переменная измеряется по порядковой шкале (ранги).
4.	Когда в выборках много выбросов или маленький размер выборок.

### Формулировка гипотез:
- H0 (нулевая гипотеза): Распределения двух выборок одинаковы.
- H1 (альтернативная гипотеза): Распределения различаются.

```python
from scipy.stats import mannwhitneyu # Импортируем нужный тест из пакета scipy

# Сравнение групп
stat, p = mannwhitneyu(Homemaker, Healthcare) # U-тест Манна-Уитни
print('Statistics=%.3f, p=%.3f' % (stat, p))


# Интерпретируем
alpha = 0.05   # Уровень значимости
if p < alpha:
    print('Одинаковые распределения (не отвергаем H0)')
else:
    print('Разные распределения (отвергаем H0)')
```

## Бутстрап или семплирование (Не бойся, это просто!)

<div class="alert alert-info">
Семплирование, простыми словами, это когда мы берем часть данных и считаем по ним метрику. И так повторяем много раз. Можем построить по метрикам гистограмму. Это экономит время, когда данных очень много.
</div>

```python
#Если вдруг оказалось сложно, то вот еще ододин способ визуализации различий
# Creating an list with bootstrapped means for each AB-group
plt.figure(figsize=(15, 10))

boot_1d = []
for i in range(1_000):
    boot_mean = data.sample(frac = 1, replace = True).groupby('version')['retention_7'].mean()  #retention_1
    boot_1d.append(boot_mean)
    
# Преобразование списка в DataFrame
boot_1d = pd.DataFrame(boot_1d)
    
# A Kernel Density Estimate plot of the bootstrap distributions
boot_1d.plot(kind='density')
plt.show()
```

<div class="alert alert-info">
<b>Задание № 1 (Обязательно):</b>
    
Профессор Оук подозревает, что покемоны в классе `grass` имеют более сильную обычную атаку, чем у покемонов в классе `rock`. Проверьте, прав ли он, и убедите его в своем выводе статистически.
    
    
Примечание: если есть покемоны, которые относятся к обоим классам, просто выбросьте их.
    
Вы можете предположить, что распределение обычных атак является нормальным для всех классов покемонов.

</div>


```python
alpha = 0.05

# 1. Отбор нужных данных
grass_attack = pokemon[pokemon['Class 1'] == 'Grass']['Attack']
rock_attack = pokemon[pokemon['Class 1'] == 'Rock']['Attack']

# 2. Проверка нормальности
if grass_attack.shape[0] < 20000:
    print('Проверка нормальности - Тест Шапиро-Уилка')
    stat, p_value_g_sh = stats.shapiro(grass_attack)
    if p_value_g_sh < alpha:
        print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: данные Grass отклоняются от нормального распределения.{bcolors.ENDC}")
    else:
        print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: данные Grass распределены нормально.{bcolors.ENDC}")

    stat, p_value_r_sh = stats.shapiro(rock_attack)
    if p_value_r_sh < alpha:
        print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: данные Rock отклоняются от нормального распределения.{bcolors.ENDC}")
    else:
        print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: данные Rock распределены нормально.{bcolors.ENDC}")
else:
    print('Проверка нормальности - Тест Колмогорова-Смирнова')
    stat, p_value_g_ks = stats.kstest(grass_attack, 'norm')
    if p_value_g_ks < alpha:
        print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: данные Grass отклоняются от нормального распределения.{bcolors.ENDC}")
    else:
        print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: данные Grass распределены нормально.{bcolors.ENDC}")

    stat, p_value_r_ks = stats.kstest(rock_attack, 'norm')
    if p_value_r_ks < alpha:
        print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: данные Rock отклоняются от нормального распределения.{bcolors.ENDC}")
    else:
        print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: данные Rock распределены нормально.{bcolors.ENDC}")

# 3. Проверка равенства дисперсий (по желанию)
print('\nПроверка равенства дисперсий - Тест Левена')
stat, p_value_l = stats.levene(grass_attack, rock_attack)
if p_value_l < alpha:
    print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: дисперсии не равны.{bcolors.ENDC}")
else:
    print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: дисперсии равны.{bcolors.ENDC}")

# 4. Проверка на нормальность
print('\nПроверка на нормальность: ', end='')
# 4.1 Если нормальность есть и дисперсии равны → t-test
if grass_attack.shape[0] < 20000 and p_value_g_sh < alpha and p_value_r_sh < alpha and p_value_l < alpha:
    print('T-test Стьюдента')
    t_stat, p_value = ttest_ind(grass_attack, rock_attack, equal_var=False)  # Лучше всегда ставить equal_var=False (тест Уэлча)
    if p_value < alpha:
        print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: средние в выборках различаются.{bcolors.ENDC}")
    else:
        print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: средние равны.{bcolors.ENDC}")
else:
    # 4.2 Если нормальности нет → критерий Манна-Уитни
    print('Тест Манна-Уитни')
    u_stat, p_value_u = mannwhitneyu(grass_attack, rock_attack, alternative='greater')
    if p_value_u < alpha:
        print(f"{bcolors.FAIL}Отвергаем нулевую гипотезу: средние в выборках различаются.{bcolors.ENDC}")
    else:
        print(f"{bcolors.OKGREEN}Не отвергаем нулевую гипотезу: средние равны.{bcolors.ENDC}")
```
