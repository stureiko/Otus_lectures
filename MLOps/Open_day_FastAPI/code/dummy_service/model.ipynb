# Модель

Примитивная модель которая будет нами использована для оборачивания в API на базе Flask и FastAPI

```python
class Model:
    def __init__(self, name='') -> None:
        self.name = name

    def fit(self, example)-> bool:
        if example == None:
            return False
        else:
            return True
        
    def predict(self, example='')-> int:
        if example == None:
            return 0
        else:
            return len(example)
        
    def put_pred(self, example)-> str:
        if example == None:
            return ''
        else:
            return f'total input {str(len(example))}'
        
```

```python
model = Model(name='dummy model')
```

```python
model.fit('test string')
```

```python
model.fit(None)
```

```python
model.predict('another string')
```

```python
model.predict('')
```

```python
model.put_pred(None)
```

```python
model.put_pred(example=None)
```
