```python
import pandas as pd
import numpy as np
from sklearn.datasets import make_blobs, make_classification
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.simplefilter("ignore", UserWarning)
```

# K-means алгоритм

```python
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=1.5, random_state=10)
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
# или если хотим раскрасить
plt.scatter(X[:, 0], X[:, 1], c=y_true)
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
X[:10]
```

```python
y_true[:10]
```

## 2 кластера

```python
kmeans = KMeans(n_clusters=2, random_state=10, n_init=10).fit(X)
```

```python
labels = kmeans.labels_
```

```python
labels
```

```python
centers = kmeans.cluster_centers_
centers
```

```python
plt.scatter(X[:,0], X[:,1], c=labels);
plt.scatter(centers[:,0], centers[:,1], marker = '*', s=200, c='r')
plt.xticks([])
plt.yticks([])
plt.show()
```

```python
kmeans_1 = KMeans(n_clusters=2, n_init=10).fit(X)
centers_1 = kmeans_1.cluster_centers_
labels_1 = kmeans_1.labels_
plt.scatter(X[:,0], X[:,1], c=labels_1);
plt.scatter(centers_1[:,0], centers_1[:,1], marker = '*', s=200, c='r');
```

## Выбор числа кластеров

```python
inertia = []
for i in range(1,11):
    kmeans = KMeans(n_clusters=i, random_state=10, n_init=10).fit(X)
    inertia_i = kmeans.inertia_
    inertia.append(inertia_i)
```

```python
inertia
```

```python
plt.plot(range(1,11), inertia, marker='o');
```

```python
D = []
for i in range(1,9):
    Di = (inertia[i] - inertia[i+1]) / (inertia[i-1] - inertia[i])
    D.append(Di)
```

```python
plt.plot(range(2,10), D, marker='o');
```

# 4. Кластеризация

## 4.1 KMeans

```python
inertia = []
for i in range(1,11):
    kmeans = KMeans(n_clusters=i, random_state=10, n_init=10).fit(X_scaled)
    labels = kmeans.labels_
    inertia_i = kmeans.inertia_
    inertia.append(inertia_i)
```

```python
plt.plot(range(1,11), inertia, marker='o');
```

```python
D = []
for i in range(1,9):
    Di = (inertia[i] - inertia[i+1])/(inertia[i-1] - inertia[i])
    D.append(Di)
```

```python
plt.plot(range(2,10), D, marker='o');
```
