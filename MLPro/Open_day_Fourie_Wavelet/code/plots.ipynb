```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python

# Параметры для нормального распределения
mean = 0  # Среднее значение
std_dev = 1  # Стандартное отклонение
n = 1000  # Количество точек во временном ряде

# Генерация нормального временного ряда
time_series = np.random.normal(mean, std_dev, n)

```

```python
# Построение графика временного ряда
plt.figure(figsize=(8, 6))
plt.plot(time_series, color='b')
plt.title("Нормальный временной ряд")
plt.xlabel("Время")
plt.ylabel("Значение")
plt.show()

```

```python
# Построение гистограммы и графика распределения значений временного ряда
plt.figure(figsize=(8, 6))
sns.histplot(time_series, kde=True, color='b')
plt.title("Распределение значений нормального временного ряда")
plt.xlabel("Значение")
plt.ylabel("Частота")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

```python
# Генерация нестационарного временного ряда с трендом
np.random.seed(0)
n = 1000
trend = np.linspace(1, 10, n)
noise = np.random.normal(0, 1, n) * trend  # Дисперсия увеличивается с трендом
non_stationary_series = trend + noise

```

```python
# График нестационарного временного ряда
plt.figure(figsize=(12, 5))
plt.plot(non_stationary_series, color='b')
plt.title("Нестационарный временной ряд")
plt.xlabel("Время")
plt.ylabel("Значение")
plt.show()

```

```python
# Применение преобразования Бокса-Кокса
transformed_series, lambda_bc = stats.boxcox(non_stationary_series - np.min(non_stationary_series) + 1)

```

```python
# График временного ряда после преобразования Бокса-Кокса
plt.figure(figsize=(12, 5))
plt.plot(transformed_series, color='g')
plt.title(f"Временной ряд после преобразования Бокса-Кокса (λ = {lambda_bc:.2f})")
plt.xlabel("Время")
plt.ylabel("Значение")
plt.show()

```

```python
# Гистограммы и графики плотности распределения до и после преобразования
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
sns.histplot(non_stationary_series, kde=True, color='b')
plt.title("Распределение значений до преобразования Бокса-Кокса")
plt.xlabel("Значение")

plt.subplot(1, 2, 2)
sns.histplot(transformed_series, kde=True, color='g')
plt.title("Распределение значений после преобразования Бокса-Кокса")
plt.xlabel("Значение")

plt.tight_layout()
plt.show()
```
