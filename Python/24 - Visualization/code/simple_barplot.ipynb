```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({'Loude': [142, 138, 134, 133, 131],
                   'Names': ['Arrowhead\nStadium,\nCansas\nCity', 'CenturyLink\nField,\nSeattle', 'Husky\nStadium,\nSeattle',
                             'Memorial\nStadium,\nClemson', 'TurkTelekom\nStadium,\nIstanbul']})
df
```

```python
plt.bar(df['Names'], df['Loude'], color='grey')
plt.ylim([130, 145])
for i, value in enumerate(df['Loude']):
    plt.text(i, value + 0.5, str(value), ha='center', fontdict={'size': 14})
# Отключаем отображение оси Y
plt.gca().axes.get_yaxis().set_visible(False)
plt.title("World's Loudest Stadiums")
plt.plot()
```
