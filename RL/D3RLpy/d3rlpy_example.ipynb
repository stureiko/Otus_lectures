# Example d3rlpy framework

using d3rlpy conda enviroment

```python
import d3rlpy
```

```python
# Import CartPole
dataset, env = d3rlpy.datasets.get_cartpole()
```

```python
dqn = d3rlpy.algos.DQNConfig().create(device='cpu')
```

```python
dqn.fit(dataset=dataset,
        n_steps=10000,
        n_steps_per_epoch=1000,
        evaluators={'enviroment': d3rlpy.metrics.EnvironmentEvaluator(env)})
```
