```python
import tensorflow as tf
```

```python
# Check if TensorFlow can access a GPU
gpu_available = tf.config.list_physical_devices('GPU')

if gpu_available:
    print("GPU is available:")
    for gpu in gpu_available:
        print(gpu)
else:
    print("No GPU available.")
```

```python

print("Результат проверки доступности GPU:")
print(f"Доступно GPU устройств: {len(gpu_available)}")
if gpu_available:
    print("Список доступных GPU:")
    for i, gpu in enumerate(gpu_available, 1):
        print(f"  {i}. {gpu}")
else:
    print("GPU устройства не найдены")
```
